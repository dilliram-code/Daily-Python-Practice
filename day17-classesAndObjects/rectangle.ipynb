```python
class Rectangle:
  def __init__(self, length):
    print("Id of r1",id(self))
    self.length = length
  
  def fun(self):
    self.breadth = 10
    
  def show(self):
    print(self.length)
    print(self.breadth)
  
  

r1 = Rectangle(5)
r1.fun()
r1.show()
print("Id of r1: ",id(r1))
```

```python
class Circle:
  
  # class variable
  count = 0
  
  def __init__(self,r):
    self.radius = r
    Circle.count += 1
  
  def area(self):
    return 3.14*self.radius*2
  
  @classmethod
  def counter(cls):
    return cls.count


circle1 = Circle(4)
circle1.counter()

circle2 = Circle(5)
circle3 = Circle(6)

Circle.counter()
```
