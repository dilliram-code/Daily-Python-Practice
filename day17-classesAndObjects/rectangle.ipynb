```python
class Rectangle:
  def __init__(self, length):
    print("Id of r1",id(self))
    self.length = length
  
  def fun(self):
    self.breadth = 10
    
  def show(self):
    print(self.length)
    print(self.breadth)
  
  

r1 = Rectangle(5)
r1.fun()
r1.show()
print("Id of r1: ",id(r1))
```

#### Class method: Example

```python
class Circle:
  
  # class variable
  count = 0
  
  def __init__(self,r):
    self.radius = r
    Circle.count += 1
  
  def area(self):
    return 3.14*self.radius*2
  
  @classmethod
  def counter(cls):
    return cls.count


circle1 = Circle(4)
circle1.counter()

circle2 = Circle(5)
circle3 = Circle(6)

Circle.counter()
```

#### Class Methods Properties:

- Declared outside all methods
- Created only once, common of all instances
- Can be accessed using object or class name
- Used as a shared data
- Stores information about class (Meta Data)

#### Static Method: 

```python
class Rectangle1:
  def __init__(self, l, b):
    self.length = l 
    self.breadth = b
  
  @staticmethod
  def calc_area(length, breadth):
    return length * breadth
  

r1 = Rectangle1(4,5)
Rectangle1.calc_area(5,6)
```

#### Static method method properties:

- utility functions 
- can be called using class name
- can't access members of a class
- @staticmethod decorator is used

#### Property Method

```python
class Rectangle2:
  def __init__(self, l, b):
    self.set_length(l)
    self.breadth = b 
  
  def get_length(self):
    return self.length
  
  def get_breadth(self):
    return self.breadth
  
  def set_length(self, l):
    if l >= 0:
      self.length = l 
    else:
      self.length = 1
  
  def set_breadth(self, b):
    if b >=0:
      self.breadth = b
    else:
      self.breadth = 1
  
r2 = Rectangle2(-4,5)
print(r2.get_length())

r2.set_breadth(-10)
print(r2.get_breadth())
```

##### Property method: getter and setter

```python
class Rectangle2:
  def __init__(self, l, b):
    self.length = l
    self.breadth = b 
  
  @property
  def length(self):
    return self._length
  
  @property
  def breadth(self):
    return self._breadth
  
  @length.setter
  def length(self, l):
    if l >= 0:
      self._length = l 
    else:
      self._length = 1
  
  @breadth.setter
  def breadth(self, b):
    if b >=0:
      self._breadth = b
    else:
      self._breadth = 1
  
r2 = Rectangle2(4,5)
r2.length = -20
r2.breadth = 100
print(r2.length)
print(r2.breadth)
```
