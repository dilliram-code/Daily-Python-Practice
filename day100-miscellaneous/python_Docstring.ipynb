#### DocString

```python
def add(a, b):
    """Add two numbers."""
    return a + b


print(add.__doc__)
```

```python
print(add.__doc__)
```

```python
# help function

def add(a, b):
    """Add two numbers and return the result."""
    return a + b


help(add)
```

##### Multiline docstring

```python
def calculate_area(length, width):
    """
    Calculate the area of a rectangle.

    Parameters:
        length: Length of the rectangle.
        width: Width of the rectangle.

    Returns:
        The area of the rectangle.
    """
    return length * width
```

```python
calculate_area(4,5)
```

##### DocString for modules

```python
# calculator.py

"""
This module provides basic mathematical operations.
"""


def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


print(__doc__)
```

##### DocString in Classes

```python
class Student:
    """Represent a student."""

    def __init__(self, name, age):
        self.name = name
        self.age = age 


print(Student.__doc__)
```

##### Methods can have DocString

```python
class Student:
    """Represent a student."""

    def __init__(self, name):
        """Initialize a student with a name."""
        self.name = name

    def introduce(self):
        """Return a short introduction."""
        return f"My name is {self.name}." 

print(Student.__doc__)
```
