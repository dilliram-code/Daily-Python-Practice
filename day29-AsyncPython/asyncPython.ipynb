##### Asynchronous Python

- Asynchronous programming allows one Python program to start an operation that is waiting for something else—such as a network response, database response, timer, or file operation—and use that waiting time to work on another operation.

```python
# first async function

import asyncio

async def task():
    print("Start")
    await asyncio.sleep(1)
    print("End")

# Just call await directly instead of asyncio.run()
await task()
```
