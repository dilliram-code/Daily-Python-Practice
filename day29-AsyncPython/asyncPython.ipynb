##### Asynchronous Python

- Asynchronous programming allows one Python program to start an operation that is waiting for something else—such as a network response, database response, timer, or file operation—and use that waiting time to work on another operation.

```python
# first async function

import asyncio

async def task():
    print("Start")
    await asyncio.sleep(1)
    print("End")

# Just call await directly instead of asyncio.run()
await task()
```

##### Structured Concurrency:

- Structured concurrency means that the lifetime of child tasks is controlled by the lifetime of their parent scope. Cancellation then becomes predictable: when the parent is cancelled or one important child fails, the related work can be cancelled together and cleaned up properly.

- This is especially important in `asyncio` because otherwise it is very easy to `create background tasks that continue running after the operation that created them has already finished.`
